```python
# cell 1
import os, sys, time
print("Python executable:", sys.executable)
print("JAVA_HOME (env):", os.environ.get("JAVA_HOME"))
print("HADOOP_HOME (env):", os.environ.get("HADOOP_HOME"))
try:
    import pyspark
    print("pyspark version:", pyspark.__version__)
    print("pyspark installed at:", pyspark.__file__)
except Exception as e:
    print("pyspark import error:", e)
```

```python
# cell 2
import time
from pyspark.sql import SparkSession

t0 = time.time()
spark = SparkSession.builder \
    .master("local[1]") \
    .appName("PysparkLearnings_Diagnostic") \
    .config("spark.ui.showConsoleProgress", "false") \
    .getOrCreate()
t1 = time.time()
print("SparkSession created in {:.2f} seconds".format(t1 - t0))

# reduce verbosity
spark.sparkContext.setLogLevel("WARN")
print("Spark master:", spark.sparkContext.master)
print("Spark app name:", spark.sparkContext.appName)
```
