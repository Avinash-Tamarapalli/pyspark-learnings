Pyspark Handling Missing Values
* Dropping Columns
* Dropping Rows
* Various Parameter in Dropping funcionalities

```python
# Code to Check the null values in a Data Frame
from pyspark.sql.functions import col, sum, when, isnan

# Calculate the count of null (or NaN for numeric types) values for each column
null_counts_df = df.select([
    sum(when(col(c).isNull() | isnan(col(c)), 1).otherwise(0)).alias(c)
    for c in df.columns
])

# Collect the result as a Row object
null_counts_row = null_counts_df.collect()[0]

# Filter out columns that have zero null values
columns_with_nulls = [
    c for c in null_counts_df.columns if null_counts_row[c] > 0
]

if columns_with_nulls:
    print("Columns with null values and their counts:")
    # Display only the columns that have null values
    null_counts_df.select(columns_with_nulls).show()
else:
    print("No columns with null values found.")
```

```python
from pyspark.sql.functions import col, sum, when, isnan

def get_null_counts(df):

  null_expressions = []
  for c in df.columns:
      null_expressions.append(sum(when(col(c).isNull() | isnan(col(c)), 1).otherwise(0)).alias(c))

  # print(null_expressions)

  null_counts_df_no_list_comp = df.select(*null_expressions)
  return null_counts_df_no_list_comp
  # null_counts_df_no_list_comp.show()
```

### `value` as a dictionary

When `value` is a dictionary, it allows you to specify different replacement values for different columns. The keys of the dictionary should be column names (strings), and the values should be the replacement values (int, float, boolean, or string).

**Important Note**: As the definition states, if `value` is a dictionary, the `subset` parameter is ignored. The dictionary itself dictates which columns to fill.

Let's say we want to fill nulls in `1980-81` with `999` and nulls in `1983-84` with the string `'Data Missing'`:

```python
print('Original null counts:')
get_null_counts(gdp).show()

fill_values = {
    '1980-81': 999,
    '1983-84': "Missing Value"
}

gdp_filled_dict = gdp.na.fill(fill_values)

print('Null counts after filling with a dictionary:')
get_null_counts(gdp_filled_dict).show()

print('DataFrame after filling with a dictionary (first 5 rows):')
gdp_filled_dict.show(5)
```

Notice in the last example that only the columns specified in the `fill_values` dictionary were affected. Any other columns with nulls (like `1981-82` to `1992-93` in the original `gdp` DataFrame) still retain their nulls because they were not mentioned in the dictionary. This demonstrates that when `value` is a dictionary, `subset` is indeed ignored, as the dictionary itself acts as the column selector for filling.

As you can see from the schema, '1983-84' is an `integer` type. Therefore, to fill its nulls, you need to provide an integer value. Try filling it with `0` (or any other appropriate integer):

Let's try filling the Null values with different strategies..
Like `mean` `median` `mode`

```python
strategy = "median"

#Filling Nulls with mean or mode. Using Imputer
from pyspark.ml.feature import Imputer

imputer = Imputer(
    inputCols = ['1980-81','1983-84'],
    outputCols= [f"{c}_imputed" for c in ['1980-81','1983-84']]
).setStrategy(strategy)

imputer.fit(gdp).transform(gdp).select(['State','1980-81','1983-84','1980-81_imputed','1983-84_imputed']).show(5)
```

```python
from pyspark.ml.feature import Imputer

wantedColumns = ['1980-81','1982-83','1983-84','1984-85']

imputer = Imputer(
    inputCols = wantedColumns,
    outputCols= [f"{c}_imputed" for c in wantedColumns]
).setStrategy("median")

filled_gdp = imputer.fit(gdp).transform(gdp).select(['State','1980-81','1982-83','1983-84','1984-85','1980-81_imputed','1982-83_imputed','1983-84_imputed','1984-85_imputed'])
filled_gdp = filled_gdp.withColumnRenamed("1980-81","1980_81").withColumnRenamed("1982-83","1982_83").withColumnRenamed("1983-84","1983_84").withColumnRenamed("1984-85","1984_85")\
.withColumnRenamed("1980-81_imputed",'1980_81_imputed').withColumnRenamed("1982-83_imputed",'1982_83_imputed').withColumnRenamed("1983-84_imputed",'1983_84_imputed').withColumnRenamed("1984-85_imputed",'1984_85_imputed')
filled_gdp.show(5)
```

Group By and Aggregate Functions

```python
df = spark.read.option("header","true").csv(content_path, inferSchema = True)
df.show(5)
```

```python
df.printSchema()
```

```python
from pyspark.sql.types import DoubleType

df = df.withColumn('Sales', df['Sales'].cast(DoubleType()))
df.printSchema()
```

```python
df.groupBy(['City']).sum('Sales').show(5)
```

```python
df.groupBy(['City','Order ID']).count().show()
```

```python
df.groupBy(['City','Order ID']).count().orderBy('City').show()
```

```python
df.groupBy(['City']).mean('Sales').alias("Mean Sales").orderBy(['City']).show(5)
```

### Option 1: Use `withColumnRenamed()` after the aggregation

This is a straightforward way to rename a column after it has been created by an aggregation.

```python
df.groupBy(['City']).mean('Sales').withColumnRenamed('avg(Sales)', 'Mean Sales').orderBy(['City']).show(5)
```

### Option 2: Use `agg()` with `alias()` on the aggregation expression

This is often considered the more idiomatic PySpark way when performing multiple aggregations, as it allows you to name your aggregated columns directly as you define the aggregations.

```python
from pyspark.sql.functions import avg

df.groupBy(['City']).agg(avg('Sales').alias('Mean Sales')).orderBy(['City']).show(5)
```

```python
df.agg({'Sales':'count'}).show(5)
```

```python
from pyspark.sql.functions import count, desc
df.groupBy(['City']).agg(count('Sales').alias('Total Sales per City')).orderBy(desc('Total Sales per City')).show(5)
```
