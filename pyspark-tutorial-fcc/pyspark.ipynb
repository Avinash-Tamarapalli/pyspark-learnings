Pyspark Handling Missing Values
* Dropping Columns
* Dropping Rows
* Various Parameter in Dropping funcionalities
* Handling Missing Values by Mean, Median and Mode

```python
# Code to Check the null values in a Data Frame
from pyspark.sql.functions import col, sum, when, isnan

# Calculate the count of null (or NaN for numeric types) values for each column
null_counts_df = df.select([
    sum(when(col(c).isNull() | isnan(col(c)), 1).otherwise(0)).alias(c)
    for c in df.columns
])

# Collect the result as a Row object
null_counts_row = null_counts_df.collect()[0]

# Filter out columns that have zero null values
columns_with_nulls = [
    c for c in null_counts_df.columns if null_counts_row[c] > 0
]

if columns_with_nulls:
    print("Columns with null values and their counts:")
    # Display only the columns that have null values
    null_counts_df.select(columns_with_nulls).show()
else:
    print("No columns with null values found.")
```

```python
from pyspark.sql.functions import col, sum, when, isnan

def get_null_counts(df):

  null_expressions = []
  for c in df.columns:
      null_expressions.append(sum(when(col(c).isNull() | isnan(col(c)), 1).otherwise(0)).alias(c))

  # print(null_expressions)

  null_counts_df_no_list_comp = df.select(*null_expressions)
  return null_counts_df_no_list_comp
  # null_counts_df_no_list_comp.show()
```

To get the dimensions of a PySpark DataFrame, you need to use two separate methods:

*   `df.count()`: Returns the number of rows in the DataFrame.
*   `len(df.columns)`: Returns the number of columns in the DataFrame.

```python
num_rows = gdp.count()
num_columns = len(gdp.columns)

print(f"Number of rows: {num_rows}")
print(f"Number of columns: {num_columns}")
```

```python
#drops the rows if any column has null
flt = gdp.na.drop(how = "any")
flt.show(5)
# It dropped all the rows where it could find null in any column
```

```python
num_rows = flt.count()
num_columns = len(gdp.columns)

print(f"Number of rows: {num_rows}")
print(f"Number of columns: {num_columns}")
```

```python
# It drops the row if every value is null
flt = gdp.na.drop(how = "all")
flt.show(5)
```

```python
num_rows = flt.count()
num_columns = len(gdp.columns)

print(f"Number of rows: {num_rows}")
print(f"Number of columns: {num_columns}")
```

```python
#The threshold indicates atleast 2(specified number) non null values needs to be available in the row
threshold = 24 #Here I specified 24 means that row should contain atleast 24 non null values or else drop the row
flt = gdp.na.drop(thresh = threshold)
# flt.show(5)
num_rows = flt.count()
num_columns = len(gdp.columns)

print(f"Number of rows: {num_rows}")
print(f"Number of columns: {num_columns}")
```
